```python
import requests
import pandas as pd

print("Downloading DOID ontology file for mapping...")
url = "https://raw.githubusercontent.com/DiseaseOntology/HumanDiseaseOntology/main/src/ontology/doid.obo"
response = requests.get(url)

# Dictionary to hold the MESH to DOID mapping
mesh_to_doid = {}
current_doid = None

# Parse the .obo file
for line in response.text.splitlines():
    if line.startswith("id: DOID:"):
        current_doid = line.replace("id: ", "").strip()
    elif line.startswith("xref: MESH:"):
        mesh_id = line.replace("xref: ", "").strip()
        if current_doid:
            mesh_to_doid[mesh_id] = current_doid

# Map the DiseaseID (formatted as MESH:XXXX) to DOID for df_cd
df_cd['DOID'] = df_cd['DiseaseID'].map(mesh_to_doid)

# Map the DiseaseID (formatted as MESH:XXXX) to DOID for df_gd
df_gd['DOID'] = df_gd['DiseaseID'].map(mesh_to_doid)

# Display the results for df_cd
print("\nDOID mapping results for df_cd:")
print(df_cd[['DiseaseName', 'DiseaseID', 'DOID']].head())

# Display the results for df_gd
print("\nDOID mapping results for df_gd:")
print(df_gd[['DiseaseName', 'DiseaseID', 'DOID']].head())
```

```python
import requests
import pandas as pd

print("Downloading DOID ontology file for MESH to UMLS mapping...")
url = "https://raw.githubusercontent.com/DiseaseOntology/HumanDiseaseOntology/main/src/ontology/doid.obo"
response = requests.get(url)

mesh_to_umls = {}
current_mesh = []
current_umls = []

# Parse the .obo file to link MESH and UMLS IDs under the same DOID terms
for line in response.text.splitlines():
    if line.startswith("[Term]"):
        # Establish mappings for the previous term
        for m in current_mesh:
            for u in current_umls:
                mesh_to_umls[m] = u
        # Reset lists for the new term
        current_mesh = []
        current_umls = []
    elif line.startswith("xref: MESH:"):
        mesh_id = line.replace("xref: ", "").strip()
        current_mesh.append(mesh_id)
    elif line.startswith("xref: UMLS_CUI:"):
        umls_id = line.replace("xref: UMLS_CUI:", "").strip()
        current_umls.append(umls_id)

# Map the final term in the file
for m in current_mesh:
    for u in current_umls:
        mesh_to_umls[m] = u

print(f"Extracted {len(mesh_to_umls)} MESH to UMLS mappings.\n")

# Apply the mapping to the DataFrames
df_cd['UMLS_CUI'] = df_cd['DiseaseID'].map(mesh_to_umls)
df_gd['UMLS_CUI'] = df_gd['DiseaseID'].map(mesh_to_umls)

# Display the results
print("UMLS mapping results for df_cd:")
print(df_cd[['DiseaseName', 'DiseaseID', 'UMLS_CUI']].head())

print("\nUMLS mapping results for df_gd:")
print(df_gd[['DiseaseName', 'DiseaseID', 'UMLS_CUI']].head())
```

```python
import requests
import pandas as pd

# Get unique MeSH IDs from all three DataFrames
unique_mesh_cd = set(df_cd['ChemicalID'].dropna().unique())
unique_mesh_cg = set(df_cg['ChemicalID'].dropna().unique())
unique_mesh_cp = set(df_cp['chemicalid'].dropna().unique())

unique_mesh = list(unique_mesh_cd | unique_mesh_cg | unique_mesh_cp)
print(f"Total unique chemicals to map across df_cd, df_cg, and df_cp: {len(unique_mesh)}")

# MyChem.info batch query url
url = 'https://mychem.info/v1/query'

mesh_to_drugbank = {}
chunk_size = 1000

print("Querying MyChem.info in batches...")
for i in range(0, len(unique_mesh), chunk_size):
    chunk = unique_mesh[i:i+chunk_size]
    # We search the unique mesh IDs against fields that commonly store MeSH IDs
    payload = {
        "q": ",".join(chunk),
        "scopes": "drugcentral.xrefs.mesh_id,pharmgkb.xref.mesh,umls.mesh,mesh",
        "fields": "drugbank.id"
    }
    try:
        res = requests.post(url, data=payload)
        if res.status_code == 200:
            results = res.json()
            for item in results:
                query_id = item.get('query')
                drugbank_data = item.get('drugbank', {})

                # Extract the DrugBank ID (can sometimes be a list if multiple are mapped)
                db_id = None
                if isinstance(drugbank_data, list):
                    db_id = drugbank_data[0].get('id') if drugbank_data else None
                elif isinstance(drugbank_data, dict):
                    db_id = drugbank_data.get('id')
                elif isinstance(drugbank_data, str): # Occasional direct string mapping
                    db_id = drugbank_data

                if db_id and query_id and query_id not in mesh_to_drugbank:
                    mesh_to_drugbank[query_id] = db_id
    except Exception as e:
        print(f"Error querying batch {i}: {e}")

# Map the extracted DrugBank IDs back to the DataFrames
df_cd['DrugBankID'] = df_cd['ChemicalID'].map(mesh_to_drugbank)
df_cg['DrugBankID'] = df_cg['ChemicalID'].map(mesh_to_drugbank)
df_cp['drugbankid'] = df_cp['chemicalid'].map(mesh_to_drugbank)

print(f"Successfully mapped {df_cd['DrugBankID'].notna().sum()} rows in df_cd.")
print(f"Successfully mapped {df_cg['DrugBankID'].notna().sum()} rows in df_cg.")
print(f"Successfully mapped {df_cp['drugbankid'].notna().sum()} rows in df_cp.")
```

```python
import pandas as pd

print("Parsing MRCONSO.RRF for GO to UMLS mappings (this might take a moment)...")

filepath = '/content/drive/MyDrive/Colab Notebooks/MRCONSO.RRF'

# Standard UMLS MRCONSO.RRF columns (the last pipe creates an empty 19th column)
columns = [
    'CUI', 'LAT', 'TS', 'LUI', 'STT', 'SUI', 'ISPREF', 'AUI', 'SAUI',
    'SCUI', 'SDUI', 'SAB', 'TTY', 'CODE', 'STR', 'SRL', 'SUPPRESS', 'CVF', 'EXTRA'
]

go_to_umls = {}

# Read the file in chunks with explicit headers to prevent high memory usage
for chunk in pd.read_csv(filepath, sep='|', header=None, names=columns, chunksize=500000, dtype=str, index_col=False):
    # Filter by Source Abbreviation (SAB) for Gene Ontology
    go_chunk = chunk[chunk['SAB'] == 'GO'].copy()

    if not go_chunk.empty:
        # Ensure the code matches the 'GO:XXXXXXX' format used in df_cp
        go_chunk['CODE'] = go_chunk['CODE'].apply(lambda x: x if str(x).startswith('GO:') else f"GO:{x}")
        # Map GO code to CUI
        chunk_dict = pd.Series(go_chunk['CUI'].values, index=go_chunk['CODE']).to_dict()
        go_to_umls.update(chunk_dict)

print(f"Extracted {len(go_to_umls)} GO to UMLS mappings.")

# Map the phenotypeid to UMLS_CUI
df_cp['UMLS_CUI'] = df_cp['phenotypeid'].map(go_to_umls)

print("\nUMLS mapping results for df_cp:")
display(df_cp[['phenotypename', 'phenotypeid', 'UMLS_CUI']].head())
print(f"\nSuccessfully mapped {df_cp['UMLS_CUI'].notna().sum()} out of {len(df_cp)} rows in df_cp.")
```

# Weight Creation

```python
query_weight = """
MATCH (c:Compound)-[r]-(d:Disease)
WHERE r.PubMedIDs IS NOT NULL
WITH r, size(r.PubMedIDs) as pmid_count
// Calculate max(1, count)
WITH r, pmid_count, CASE WHEN pmid_count > 1 THEN pmid_count ELSE 1 END as max_val
// Calculate ln(max_val), handling the log(0) edge case by just setting weight to 0
SET r.weight = CASE WHEN max_val > 0 THEN log(max_val) ELSE 0.0 END
RETURN count(r) as updated_count
"""

try:
    with GraphDatabase.driver(uri, auth=(user, password)) as driver:
        with driver.session() as session:
            result = session.run(query_weight)
            count = result.single()["updated_count"]
            print(f"Successfully added weight to {count} relationships.")
except Exception as e:
    print(f"An error occurred: {e}")
```

This code cell sets up a local graph analysis environment to evaluate how effectively random walks can predict drug-disease connections.

First, it fetches graph data from Neo4j, pulling in Compounds, Diseases, and their relationship weights (derived from PubMed evidence). It uses the NetworkX library to build a local version of this graph.

Second, it defines a 'simulate_random_walks' function. This function mimics a process where you start at a drug and 'step' through the graph. You can choose to walk 'unweighted' (stepping randomly) or 'weighted' (preferring paths with more scientific evidence).

Finally, the 'evaluate_walks' function tests the accuracy of these walks. It takes a sample of known drug-disease pairs, starts a walk at the drug, and sees if the walk successfully 'finds' the correct disease. It calculates a 'Hits@10' score, which tells us how often the true disease appears in the top 10 most-visited nodes. This allows us to quantitatively compare if using weighted evidence actually improves our ability to recover known medical relationships.

```python
import networkx as nx
import random
from collections import Counter

# Fetch the graph data
query_edges = """
MATCH (d:Disease)-[r]-(c:Compound)
RETURN c.name AS compound, d.name AS disease, coalesce(r.weight, 0.1) AS weight
"""

G = nx.Graph()
test_pairs = []

print("Fetching graph data...")
with GraphDatabase.driver(uri, auth=(user, password)) as driver:
    with driver.session() as session:
        result = session.run(query_edges)
        for record in result:
            c = record['compound']
            d = record['disease']
            w = record['weight']
            # Add nodes with bipartite attribute for clarity
            G.add_node(c, type='Compound')
            G.add_node(d, type='Disease')
            # For unweighted, we will just use weight=1.0 later, but store the real weight here
            G.add_edge(c, d, weight=w)
            test_pairs.append((c, d))

print(f"Graph built with {G.number_of_nodes()} nodes and {G.number_of_edges()} edges.")

# Define the Random Walk function
def simulate_random_walks(G, start_node, num_walks=50, walk_length=5, use_weights=False):
    visits = Counter()
    for _ in range(num_walks):
        curr = start_node
        for _ in range(walk_length):
            neighbors = list(G.neighbors(curr))
            if not neighbors:
                break
            if use_weights:
                # Add a small epsilon (0.01) to prevent the sum of weights from being exactly 0
                weights = [max(0.01, G[curr][nbr].get('weight', 0.1)) for nbr in neighbors]
                curr = random.choices(neighbors, weights=weights, k=1)[0]
            else:
                curr = random.choice(neighbors)

            if G.nodes[curr]['type'] == 'Disease':
                visits[curr] += 1
    return visits

# Evaluate Hits and Hits@10
def evaluate_walks(use_weights):
    hits = 0
    hits_at_10 = 0

    # To speed up, we'll evaluate a random sample of 100 known pairs
    sample_pairs = random.sample(test_pairs, min(100, len(test_pairs)))

    for c, true_d in sample_pairs:
        visits = simulate_random_walks(G, c, num_walks=100, walk_length=4, use_weights=use_weights)

        if not visits:
            continue

        # Rank diseases by visit count
        ranked_diseases = [node for node, count in visits.most_common()]

        if true_d in ranked_diseases:
            hits += 1
            if ranked_diseases.index(true_d) < 10:
                hits_at_10 += 1

    return hits, hits_at_10, len(sample_pairs)

print("\n--- Unweighted Random Walks ---")
u_hits, u_hits10, total_samples = evaluate_walks(use_weights=False)
print(f"Successful Hits: {u_hits} / {total_samples}")
print(f"Hits@10: {u_hits10} / {total_samples} ({(u_hits10/total_samples)*100:.2f}%)")

print("\n--- Weighted Random Walks ---")
w_hits, w_hits10, total_samples = evaluate_walks(use_weights=True)
print(f"Successful Hits: {w_hits} / {total_samples}")
print(f"Hits@10: {w_hits10} / {total_samples} ({(w_hits10/total_samples)*100:.2f}%)")
```

This code is designed to test how effectively different random walk strategies can 'find' a specific target drug starting from a disease node.

First, it defines a function specific_random_walk that takes a start node and simulates multiple walks through the graph. The key feature here is the use_weights toggle: if enabled, the walker prefers edges with higher weights (derived from PubMed evidence); if disabled, it chooses neighbors at random.

Second, the code runs this simulation for both an 'Unweighted' and a 'Weighted' walk, starting at 'Type 2 Diabetes Mellitus' and aiming for 'Metformin'.

Finally, it evaluates the performance using a 'Hits@10' metric. This check determines if Metformin is among the top 10 most-visited compounds during the walks and identifies its specific rank. This allows us to see if incorporating scientific evidence (weights) makes the target drug more 'visible' in the graph traversal compared to a purely random approach.

```python
import random
from collections import Counter

start_node = "type 2 diabetes mellitus"
target_node = "Metformin"

def specific_random_walk(G, start, num_walks=1000, walk_length=4, use_weights=False):
    visits = Counter()
    for _ in range(num_walks):
        curr = start
        for _ in range(walk_length):
            neighbors = list(G.neighbors(curr))
            if not neighbors:
                break

            if use_weights:
                weights = [max(0.01, G[curr][nbr].get('weight', 0.1)) for nbr in neighbors]
                curr = random.choices(neighbors, weights=weights, k=1)[0]
            else:
                curr = random.choice(neighbors)

            # We want to hit Metformin, which is a Compound
            if G.nodes[curr]['type'] == 'Compound':
                visits[curr] += 1
    return visits

print(f"Targeting '{target_node}' from '{start_node}'\n")

for walk_type, is_weighted in [("Unweighted", False), ("Weighted", True)]:
    visits = specific_random_walk(G, start_node, num_walks=1000, walk_length=4, use_weights=is_weighted)

    # Rank compounds by visit count
    ranked_compounds = [node for node, count in visits.most_common()]

    hit = target_node in ranked_compounds

    # Check if hit is in top 10
    hit_at_10 = False
    rank = -1
    if hit:
        rank = ranked_compounds.index(target_node) + 1
        if rank <= 10:
            hit_at_10 = True

    print(f"--- {walk_type} Random Walk ---")
    print(f"Successful Hit: {'Yes' if hit else 'No'}")
    print(f"Hits@10: {'Yes' if hit_at_10 else 'No'} (Rank: {rank if hit else 'N/A'})")
    print(f"Total visits to Metformin: {visits[target_node]}")
    print("-" * 30)
```

This code finds the most direct connections between a specific drug ('Pioglitazone') and a disease ('type 2 diabetes mellitus') using two different methods. First, it searches for an 'unweighted' path, which simply finds the route with the fewest steps or 'hops' between the two nodes. Second, it searches for a 'weighted' path by calculating the total scientific evidence weight along the routes (up to 4 hops) and selecting the path with the highest cumulative weight. It then executes these queries and prints both paths along with their scores for comparison.

```python
compound_name = 'Pioglitazone'
disease_name = 'type 2 diabetes mellitus'

# Unweighted shortest path using built-in shortestPath function
query_unweighted = """
MATCH p=shortestPath((c:Compound {name: $c_name})-[*..5]-(d:Disease {name: $d_name}))
RETURN [n in nodes(p) | coalesce(n.name, n.id)] AS path, length(p) AS cost
"""

# Weighted shortest path by calculating path weights
# Limiting depth to 4 to prevent combinatorial explosion on AuraDB free tier
query_weighted = """
MATCH p=(c:Compound {name: $c_name})-[r*1..4]-(d:Disease {name: $d_name})
WITH p, reduce(cost = 0.0, rel in r | cost + coalesce(rel.weight, 1.0)) AS total_weight
RETURN [n in nodes(p) | coalesce(n.name, n.id)] AS path, total_weight AS cost
ORDER BY total_weight DESC
LIMIT 1
"""

try:
    with GraphDatabase.driver(uri, auth=(user, password)) as driver:
        with driver.session() as session:
            print("--- Unweighted Shortest Path ---")
            res_unweighted = session.run(query_unweighted, c_name=compound_name, d_name=disease_name).single()
            if res_unweighted:
                print(f"Path: {' -> '.join(res_unweighted['path'])}")
                print(f"Cost (Hops): {res_unweighted['cost']}")
            else:
                print("No unweighted path found.")

            print("\n--- Weighted Shortest Path ---")
            res_weighted = session.run(query_weighted, c_name=compound_name, d_name=disease_name).single()
            if res_weighted:
                print(f"Path: {' -> '.join(res_weighted['path'])}")
                print(f"Cost (Total Weight): {res_weighted['cost']:.4f}")
            else:
                print("No weighted path found.")
except Exception as e:
    print(f"An error occurred: {e}")
```

# Graph Connectedness

This code identifies densely connected pairs of drugs and diseases within the graph.

First, it defines a Cypher query that specifically looks for Compounds and Diseases connected by paths of exactly 3 'hops'. It groups these paths by their start and end nodes, counts them, and filters out any pairs that have 2 or fewer paths between them. For the remaining highly-connected pairs, it also calculates the absolute shortest path length between them, sorting the results to show the top 10 most connected pairs.

Second, the Python script connects to the Neo4j database, runs this query, and loops through the results, printing a formatted statement for each pair that includes the start node, end node, the number of 3-hop paths connecting them, and the length of their shortest possible path.

```python
query_multiple_paths = """
MATCH p = (start_node:Compound)-[*3]-(end_node:Disease)
WITH start_node, end_node, count(p) AS path_count
WHERE path_count > 2
RETURN start_node.name AS start, end_node.name AS end, path_count, length(shortestPath((start_node)-[*]-(end_node))) AS shortest_length
ORDER BY path_count DESC
LIMIT 10
"""

print("Looking for pairs of nodes with multiple paths between them...")
try:
    with GraphDatabase.driver(uri, auth=(user, password)) as driver:
        with driver.session() as session:
            result = session.run(query_multiple_paths)
            found = False
            for record in result:
                found = True
                print(f"Start Node: {record['start']} | End Node: {record['end']} | Number of Paths: {record['path_count']} | Shortest Path Length: {record['shortest_length']}")
            if not found:
                print("No densely connected pairs found with the given criteria.")
except Exception as e:
    print(f"An error occurred: {e}")
```

This code snippet sets up the foundation for building a local network graph. First, it defines a Cypher query (query_edges) that retrieves all pairs of connected nodes from the Neo4j database, fetching their names, labels, and the weight of their connection. A key detail here is the use of coalesce, which ensures that if an edge doesn't have a weight assigned, it defaults to a very small value of 0.01 instead of breaking. Finally, it initializes an empty graph G using the NetworkX library and prints a status message indicating the build process has started.

```python
import networkx as nx
import random
from collections import Counter

# Re-fetch graph to ensure it's available and assign small weights to unweighted relationships
query_edges = """
MATCH (n)-[r]-(m)
// coalesce ensures a small value (0.01) is used if weight is NULL
RETURN coalesce(n.name, n.id) AS source, labels(n)[0] AS source_type,
       coalesce(m.name, m.id) AS target, labels(m)[0] AS target_type,
       coalesce(r.weight, 0.01) AS weight
"""

G = nx.Graph()
print("Building local Graph...")
try:
    with GraphDatabase.driver(uri, auth=(user, password)) as driver:
        with driver.session() as session:
            result = session.run(query_edges)
            for record in result:
                s = record['source']
                t = record['target']
                w = record['weight']
                s_type = record['source_type']
                t_type = record['target_type']
                G.add_node(s, type=s_type)
                G.add_node(t, type=t_type)
                G.add_edge(s, t, weight=w)
    print(f"Graph built with {G.number_of_nodes()} nodes and {G.number_of_edges()} edges.")
except Exception as e:
    print(f"Database error: {e}")

def run_weighted_random_walks(G, start_node, num_walks=1000, walk_length=5):
    visits = Counter()
    for _ in range(num_walks):
        curr = start_node
        for _ in range(walk_length):
            neighbors = list(G.neighbors(curr))
            if not neighbors:
                break

            # Ensure weight is at least 0.01 as requested
            weights = [max(0.01, G[curr][nbr].get('weight', 0.01)) for nbr in neighbors]
            curr = random.choices(neighbors, weights=weights, k=1)[0]

            # Record disease visits
            if G.nodes[curr]['type'] == 'Disease' and curr != start_node:
                visits[curr] += 1
    return visits

print("\n--- Random Walk Results ---")
# Use top_compounds_list from the previous cell
for cmpd in top_compounds_list:
    if cmpd not in G:
        print(f"\nCompound '{cmpd}' not found in the graph.")
        continue

    # Get ground truth (actual linked diseases)
    true_diseases = set([n for n in G.neighbors(cmpd) if G.nodes[n]['type'] == 'Disease'])

    # Perform walks
    visits = run_weighted_random_walks(G, cmpd, num_walks=1000, walk_length=4)

    # Rank and evaluate
    ranked_diseases = [node for node, count in visits.most_common()]
    top_10 = ranked_diseases[:10]

    hits = len([d for d in top_10 if d in true_diseases])
    hit_at_10 = hits > 0

    print(f"\nCompound: {cmpd}")
    print(f"Known Connected Diseases: {len(true_diseases)}")
    print(f"Hits@10: {'Yes' if hit_at_10 else 'No'} ({hits} true diseases found in top 10 ranked nodes)")
    print(f"Top 3 recommended: {', '.join(top_10[:3])}")
```

## Lookahead Random Walk

Function for a random walk that considers both immediate and next-neighbor edge weights.


```python
def run_lookahead_walks(G, start_node, num_walks=1000, walk_length=4):
    visits = Counter()
    for _ in range(num_walks):
        curr = start_node
        for _ in range(walk_length):
            neighbors = list(G.neighbors(curr))
            if not neighbors:
                break

            lookahead_weights = []
            for nbr in neighbors:
                imm_weight = max(0.01, G[curr][nbr].get('weight', 0.01))
                next_neighbors = list(G.neighbors(nbr))
                next_weight_sum = sum([max(0.01, G[nbr][nn].get('weight', 0.01)) for nn in next_neighbors])
                lookahead_weights.append(imm_weight + next_weight_sum)

            curr = random.choices(neighbors, weights=lookahead_weights, k=1)[0]

            if G.nodes[curr]['type'] == 'Disease' and curr != start_node:
                visits[curr] += 1
    return visits
```

## Progressive Weight Walk

Custom random walk function where the probability of choosing a neighbor transitions from unweighted (equal probability) at the first step to fully weighted by the last step of the walk.


### Progressive Weight Walk Algorithm: Step-by-Step

1. **Initialization:** Start with a global counter to keep track of visit frequencies for target nodes (e.g., Diseases).
2. **Iterate Walks:** For each walk (up to `num_walks`), set the starting node as the current node.
3. **Traverse Steps:** For each step in the walk (from `0` to `walk_length - 1`):
    * **Get Neighbors:** Retrieve all adjacent nodes. Stop the walk early if a dead end is reached.
    * **Calculate Interpolation Factor (Alpha):** Compute `alpha = step / (walk_length - 1)`. This creates a sliding scale from `0.0` (first step) to `1.0` (last step).
    * **Determine Unweighted Probability:** Calculate the equal-chance probability for any neighbor (`1.0 / total_neighbors`).
    * **Determine Weighted Probability:** Retrieve the baseline edge weights (applying a minimum threshold like `0.01`). Normalize these weights by dividing each by the sum of all neighbor weights.
    * **Compute Final Probability:** Combine the probabilities using alpha: `(1 - alpha) * unweighted_probability + alpha * weighted_probability`. This forces early steps to be exploratory (random) and later steps to be exploitative (following evidence).
    * **Move:** Randomly select the next node using this interpolated probability distribution.
    * **Record Visit:** If the selected node matches the target criteria (e.g., it is a Disease node and not the starting node), increment its tally in the global counter.
4. **Return Results:** After all walks are finished, return the global counter with the aggregated visit counts.

```python
def run_progressive_walks(G, start_node, num_walks=1000, walk_length=4):
    visits = Counter()
    for _ in range(num_walks):
        curr = start_node
        for step in range(walk_length):
            neighbors = list(G.neighbors(curr))
            if not neighbors:
                break

            alpha = step / max(1, (walk_length - 1))

            weights = [max(0.01, G[curr][nbr].get('weight', 0.01)) for nbr in neighbors]
            total_weight = sum(weights)

            p_unweighted = 1.0 / len(neighbors)

            final_probs = []
            for w in weights:
                p_weighted = w / total_weight
                p_final = (1 - alpha) * p_unweighted + alpha * p_weighted
                final_probs.append(p_final)

            curr = random.choices(neighbors, weights=final_probs, k=1)[0]

            if G.nodes[curr].get('type') == 'Disease' and curr != start_node:
                visits[curr] += 1

    return visits
```

## Dynamic Weight Walk

Dynamic weight random walk function where chosen edge weights are temporarily increased during the walk. Execute this walk starting from 'Bortezomib' and rank the diseases found within 2 to 5 hops, evaluating how this dynamic weighting approach performs compared to the previous methods.


### Dynamic Weight Walk Algorithm: Step-by-Step

1. **Initialization:** Start with a global counter to keep track of the visit frequencies for target nodes (e.g., Diseases).
2. **Iterate Walks:** For each walk (up to `num_walks`):
    * Set the starting node as the current node.
    * Initialize a local memory state (e.g., a dictionary) to track modified edge weights specifically for this current walk.
3. **Traverse Steps:** For each step in the walk (up to `walk_length`):
    * **Get Neighbors:** Retrieve all adjacent nodes to the current node. Stop the walk early if it hits a dead end (no neighbors).
    * **Evaluate Edge Weights:** For each connecting edge, check if a modified weight exists in the local memory. If not, fetch its original baseline weight from the graph (applying a minimum threshold like `0.01`).
    * **Select Next Node:** Randomly pick the next node using these evaluated weights as the transition probability distribution.
    * **Apply Dynamic Multiplier:** Multiply the weight of the freshly traversed edge by a defined `multiplier` (e.g., `2.0`) and update it in the local memory to temporarily reinforce this path for the remainder of the current walk.
    * **Move:** Update the current node to the selected next node.
    * **Record Visit:** If the current node matches the target criteria (e.g., it is a Disease node and not the starting node), increment its tally in the global visit counter.
4. **Return Results:** After completing all specified walks, return the global counter containing the aggregated visit counts.

**Algorithm 1 The Dynamic Weight Walk technique's pseudo-code**
Initialize global visits counter
for walk = 1 : num_walks do
    curr = start_node
    Initialize local memory dynamic_weights
    for step = 1 : walk_length do
        neighbors = neighbors of curr
        if neighbors is empty then
            break
        end if
        for each nbr in neighbors do
            if edge(curr, nbr) is not in dynamic_weights then
                dynamic_weights[edge(curr, nbr)] = max(0.01, baseline_weight)
            end if
        end for
        next_node = Random selection from neighbors using dynamic_weights as probabilities
        dynamic_weights[edge(curr, next_node)] = dynamic_weights[edge(curr, next_node)] * multiplier
        curr = next_node
        if curr is of type 'Disease' and curr != start_node then
            Increment visits[curr]
        end if
    end for
end for
Return visits

```python
def run_dynamic_weight_walks(G, start_node, num_walks=1000, walk_length=4, multiplier=2.0):
    visits = Counter()
    for _ in range(num_walks):
        curr = start_node
        dynamic_weights = {}
        for _ in range(walk_length):
            neighbors = list(G.neighbors(curr))
            if not neighbors:
                break

            current_weights = []
            for nbr in neighbors:
                edge = tuple(sorted((curr, nbr)))
                if edge not in dynamic_weights:
                    dynamic_weights[edge] = max(0.01, G[curr][nbr].get('weight', 0.01))
                current_weights.append(dynamic_weights[edge])

            next_node = random.choices(neighbors, weights=current_weights, k=1)[0]

            # Update the weight of the traversed edge
            traversed_edge = tuple(sorted((curr, next_node)))
            dynamic_weights[traversed_edge] *= multiplier

            curr = next_node

            if G.nodes[curr].get('type') == 'Disease' and curr != start_node:
                visits[curr] += 1

    return visits
```

## Final Summary of Algorithms

Throughout this analysis, we have implemented and compared several graph traversal algorithms to explore relationships between **Compounds** and **Diseases**:

1.  **Standard Unweighted Random Walk**:
    *   **Logic**: Equal probability for all neighbors at every step.
    *   **Goal**: Establish a baseline for graph connectivity without bias.

2.  **Standard Weighted Random Walk**:
    *   **Logic**: Uses weights (log-normalized PubMedID counts) to bias transitions.
    *   **Goal**: Prioritize paths with stronger evidentiary support in medical literature.

3.  **Lookahead Random Walk**:
    *   **Logic**: Transition probabilities are calculated based on the weight of the immediate edge plus the sum of weights of the next-neighbors' edges.
    *   **Goal**: Identify 'hubs' and look deeper into the neighborhood, though it occasionally led to trapping in high-degree nodes.

4.  **Progressive Weight Random Walk**:
    *   **Logic**: Dynamically transitions from unweighted probabilities (alpha=0) at the start of the walk to fully weighted probabilities (alpha=1) at the final step.
    *   **Goal**: Balance broad initial exploration with targeted exploitation of weights for distant nodes.

5.  **Dynamic Weight Random Walk**:
    *   **Logic**: Temporarily increases the weight of an edge (using a multiplier) as soon as it is traversed within a single walk.
    *   **Goal**: Reinforce specific paths and discover reinforced sub-graph clusters.

6.  **Shortest Path Analysis (Unweighted vs. Weighted Cost)**:
    *   **Logic**: Used Dijkstra's algorithm where 'cost' is defined as the inverse of the edge weight.
    *   **Goal**: Directly compare the most efficient hop-based path vs. the most 'evidentiary-heavy' path.

Across these experiments, we used several algorithmic approaches to navigate the drug-disease network. Here is a breakdown of how each function makes decisions:

1. Standard Weighted Random Walk (run_walks)
Decision Logic: At each node $i$$i$, the probability of moving to neighbor $j$$j$ is proportional to the edge weight $w_{ij}$$w_{ij}$ (derived from PubMed evidence).
Formula: $P(i o j) = rac{w_{ij}}{\sum w_{ik}}$$P(i o j) = rac{w_{ij}}{\sum w_{ik}}$.
Behavior: Exploitative. It sticks closely to well-documented medical relationships.
2. Lookahead Random Walk (run_lookahead_walks)
Decision Logic: Instead of just looking at the next step, it evaluates the 'quality' of the neighborhood it is entering. It sums the weights of the edges leaving each neighbor.
Formula: $P(i o j) \propto w_{ij} + \sum_{k \in N(j)} w_{jk}$$P(i o j) \propto w_{ij} + \sum_{k \in N(j)} w_{jk}$.
Behavior: Hub-seeking. It is strongly attracted to 'hubs' (nodes with many connections), which can sometimes lead to the walker getting 'trapped' in dense clusters of compounds.
3. Progressive Weight Random Walk (run_progressive_walks)
Decision Logic: It uses a linear interpolation factor $\alpha$$\alpha$ that increases with the step number. Early steps are random (exploratory), while later steps follow the evidence (exploitative).
Formula: $P_{final} = (1 - \alpha) P_{random} + \alpha P_{weighted}$$P_{final} = (1 - \alpha) P_{random} + \alpha P_{weighted}$.
Behavior: Balanced. It allows the walker to escape the immediate 'neighborhood' of the starting drug before narrowing its search to find distant, relevant diseases.
4. Dynamic Weight Random Walk (run_dynamic_weight_walks)
Decision Logic: Within a single walk, if an edge is traversed, its weight is temporarily increased (multiplied by 2.0).
Behavior: Reinforcing. It mimics 'path-following' behavior, effectively strengthening specific routes to see which clusters are most reachable through repeated reinforcement.
